# Aufgabe 1

```python
anfangsbestand = 1080
endbestand = 1160
einstandspreis = 1000
jahresverbrauch = 4000
marktzins = 0.1
```

```python
#low-quality but works
class Formatierer:
    def formatieren(self, präfix: str, wert: float, einheit: str = '', nachkommastellen: int = 2) -> str:
        if einheit == '%':
            if type(wert) == int:
                raise TypeError('Einheit % fordert Typ float bei wert')
            return f"{präfix}: {self.prozent_formatieren(wert, nachkommastellen)}"
        else:
            return f"{präfix}: {self.tausend_formatieren(float(wert), nachkommastellen)} {einheit}"

    def tausend_formatieren(self, wert: float, nachkommastellen: int) -> str:
        formatierter_wert = f"{round(wert, nachkommastellen):.{nachkommastellen}f}"
        
        if '.' in formatierter_wert:
            ganzzahl, nachkomma = formatierter_wert.split('.')
            ganzzahl_mit_punkten = f"{int(ganzzahl):,}".replace(',', '.')
            return f"{ganzzahl_mit_punkten},{nachkomma}"
        else:
            return f"{int(formatierter_wert):,}".replace(',', '.')

    def prozent_formatieren(self, wert: float, nachkommastellen: int) -> str:
        return f"{round(wert*100, nachkommastellen):.{nachkommastellen}f} %".replace('.', ',')

f = Formatierer()
```

---

## 1. Durchschnittlicher Lagerbestand

Den durchschnittlichen Lagerbestand berechnen wir aus den beiden gegebenen Werten. Im konkreten Fall also:

$\text{Lagerbestand} = \frac{1080 + 1160}{2} = 1\,120 \text{ t}$

```python
lagerbestand = (anfangsbestand + endbestand) / 2
print(f.formatieren('Durchschnittlicher Lagerbestand', lagerbestand, 't', 0))
```

---

## 2. Umschlagshäufigkeit

Die Umschlagshäufigkeit berechnet sich aus dem Jahresverbrauch:

$UH = \frac{4000}{1120} \approx 3{,}57$

```python
uh = jahresverbrauch / lagerbestand
print(f.formatieren('Umschlagshäufigkeit', uh))
```

---

## 3. Durchschnittliche Lagerdauer

Als nächstes brauchen wir die durchschnittliche Lagerdauer. Diese wird mit $360$ dividiert durch die Umschlagshäufigkeit berechnet:

$\text{Lagerdauer} = \frac{360}{4000/1120} = 100{,}80 \text{ Tage}$

```python
lagerdauer = 360 / uh
print(f.formatieren('Durchschnittliche Lagerdauer',lagerdauer,'Tage'))
```

---

## 4. Lagerzinssatz

Als nächstes der Lagerzinssatz (Zusammenhang von Zinssatz und Lagerdauer):

$\text{Lagerzinssatz} = \frac{10\% \cdot 100{,}8}{360} = 2{,}80\%$

```python
lagerzinssatz = marktzins * lagerdauer / 360
print(f.formatieren('Lagerzinssatz',lagerzinssatz,'%'))
```

---

## 5. Lagerwert

Der Lagerwert ergibt sich aus dem Einstandspreis multipliziert mit dem durchschnittlichen Lagerbestand:

$\text{Lagerwert} = 1000 \cdot 1120 = 1.120.000 \text{ EUR}$

```python
lagerwert = einstandspreis * lagerbestand
print(f.formatieren('Lagerwert',lagerwert,'€'))
```

---

## 6. Lagerzinsen

Lagerzinsen werden nach folgender Formel berechnet:

$\text{Lagerzinsen} = \text{Lagerwert} \cdot \text{Lagerzinssatz}$

Im konkreten Fall:

$\text{Lagerzinsen} = 1.120.000 \cdot 2{,}80\% = 31.360{,}00 \text{ EUR}$

```python
lagerzinsen = lagerwert * lagerzinssatz

print(f.formatieren('Lagerzinsen', lagerzinsen, '€'))
```

## 7. Änderungen bei der Anwendung von JIT

Was passiert bei JIT (Just-in-Time)?

| Kennzahl | Effekt | Begründung |
|----------|--------|------------|
| Lagerbestand | $\downarrow$ sinkt | Weniger Lagerware |
| Lagerdauer | $\downarrow$ sinkt | Waren bleiben kürzere Zeit im Lager |
| Lagerzinsen | $\downarrow$ sinken | Weniger gebundenes Kapital |
| Umschlagshäufigkeit | $\uparrow$ steigt | Höhere Umlaufrate |
